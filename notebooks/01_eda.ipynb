```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# ============================================================
# 1. LOAD DATASET
# ============================================================

PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "heart_disease_uci.csv"

df = pd.read_csv(DATA_PATH)

print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 2. DATA INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("DATA TYPES AND NON-NULL COUNTS")
print("=" * 60)

df.info()


# ============================================================
# 3. SUMMARY STATISTICS
# ============================================================

print("\n" + "=" * 60)
print("SUMMARY STATISTICS")
print("=" * 60)

display(df.describe())


# ============================================================
# 4. MISSING VALUES
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)

print(df.isnull().sum())


# ============================================================
# 5. DUPLICATE ROWS
# ============================================================

print("\n" + "=" * 60)
print("DUPLICATE ROWS")
print("=" * 60)

print("Duplicate rows:", df.duplicated().sum())


# ============================================================
# 6. ORIGINAL TARGET DISTRIBUTION
# ============================================================

print("\n" + "=" * 60)
print("ORIGINAL TARGET DISTRIBUTION")
print("=" * 60)

print(df["num"].value_counts().sort_index())


# ============================================================
# 7. CREATE CLEAN WORKING DATAFRAME
# ============================================================

df_clean = df.copy()

# Convert original target:
# 0 = No heart disease
# 1-4 = Heart disease present

df_clean["target"] = (df_clean["num"] > 0).astype(int)

# Remove original target to prevent target leakage
df_clean = df_clean.drop(columns=["num"])

print("\nCleaned dataset shape:", df_clean.shape)

print("\nBinary target distribution:")
print(df_clean["target"].value_counts().sort_index())


# ============================================================
# 8. CLASS BALANCE
# ============================================================

class_counts = df_clean["target"].value_counts().sort_index()

class_percentages = (
    df_clean["target"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

print("\n" + "=" * 60)
print("CLASS BALANCE")
print("=" * 60)

print("\nClass counts:")
print(class_counts)

print("\nClass percentages:")
print(class_percentages.round(2))


# ============================================================
# 9. CLASS BALANCE VISUALIZATION
# ============================================================

plt.figure(figsize=(7, 5))

plt.bar(
    ["No Heart Disease", "Heart Disease"],
    class_counts.values
)

plt.title("Heart Disease Class Distribution")
plt.xlabel("Target Class")
plt.ylabel("Number of Patients")

plt.tight_layout()

plt.savefig(
    "../screenshots/class_balance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 10. DEFINE NUMERICAL AND CATEGORICAL FEATURES
# ============================================================

numerical_features = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]

categorical_features = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]

print("\n" + "=" * 60)
print("FEATURE TYPES")
print("=" * 60)

print("\nNumerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)


# ============================================================
# 11. CATEGORICAL FEATURE VALUES
# ============================================================

print("\n" + "=" * 60)
print("CATEGORICAL FEATURE VALUES")
print("=" * 60)

for column in categorical_features:
    values = sorted(df_clean[column].dropna().unique())
    print(f"{column}: {values}")


# ============================================================
# 12. NUMERICAL FEATURE HISTOGRAMS
# ============================================================

df_clean[numerical_features].hist(
    figsize=(12, 8),
    bins=20,
    edgecolor="black"
)

plt.suptitle(
    "Distribution of Numerical Features",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    "../screenshots/numerical_histograms.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 13. CORRELATION MATRIX
# ============================================================

correlation_matrix = df_clean.corr(numeric_only=True)


# ============================================================
# 14. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Heart Disease Features")

plt.tight_layout()

plt.savefig(
    "../screenshots/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 15. TARGET CORRELATIONS
# ============================================================

target_correlations = (
    correlation_matrix["target"]
    .drop("target")
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("TARGET CORRELATIONS")
print("=" * 60)

print(target_correlations)


# ============================================================
# 16. ABSOLUTE TARGET CORRELATIONS
# ============================================================

target_correlations_absolute = (
    correlation_matrix["target"]
    .drop("target")
    .abs()
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("ABSOLUTE TARGET CORRELATIONS")
print("=" * 60)

print(target_correlations_absolute)


# ============================================================
# 17. SAVE PROCESSED DATASET
# ============================================================

PROCESSED_DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "heart_disease_processed.csv"
)

df_clean.to_csv(
    PROCESSED_DATA_PATH,
    index=False
)

print("\nProcessed dataset saved to:")
print(PROCESSED_DATA_PATH)


# ============================================================
# 18. FINAL VERIFICATION
# ============================================================

print("\n" + "=" * 60)
print("FINAL VERIFICATION")
print("=" * 60)

print("\nFinal dataset shape:")
print(df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())

print("\nFinal target distribution:")
print(df_clean["target"].value_counts().sort_index())

print("\nEDA COMPLETED SUCCESSFULLY")
```
